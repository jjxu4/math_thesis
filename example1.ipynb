```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from solver import solve_vlasov_poisson
```

```python
x_res = 2**11
v_res = 2**11
x = np.linspace(0.0, 4.0 * np.pi, x_res)
v = np.linspace(-6.0, 6.0, v_res)
X, V = np.meshgrid(x, v)

a = 0.2
b = 2.0 * np.pi

chi = np.exp(-a * (X - b)**2) * np.sin(X / 2.0)**2
f0 = chi * np.exp(-V**2 / 2.0) / (2.0 * np.pi)
H = np.zeros_like(x)
```

```python
T = 20.0
dt = 0.01
t_res = int(round(T / dt))

times, f_history = solve_vlasov_poisson(f0, x, v, T, t_res, H)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
mesh = ax.pcolormesh(x, v, f_history[0], shading="auto", cmap="viridis", vmin=0.0, vmax=f0.max())
ax.set_xlim(x[0], x[-1])
ax.set_ylim(v[0], v[-1])
ax.set_xlabel("x")
ax.set_ylabel("v")
ax.set_title(f"t = {times[0]:.1f}")
fig.colorbar(mesh, ax=ax, label="f(x, v)")
fig.tight_layout()

def update(frame):
    mesh.set_array(f_history[frame].ravel())
    ax.set_title(f"t = {times[frame]:.1f}")
    return mesh,

anim = FuncAnimation(fig, update, frames=len(times), interval=100)
plt.close(fig)
HTML(anim.to_jshtml())
```
