```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from solver import solve_vlasov_poisson, plot_vp_solution
from optimizer_2 import optimizer
```

```python
x_res = 2**7
v_res = 2**7
x = np.linspace(0.0, 4.0 * np.pi, x_res)
v = np.linspace(-6.0, 6.0, v_res)
X, V = np.meshgrid(x, v)

a = 0.2
b = 2.0 * np.pi

chi = np.exp(-a * (X - b)**2) * np.sin(X / 2.0)**2
f0 = chi * np.exp(-V**2 / 2.0) / np.sqrt(2.0 * np.pi)
f_desired = f0.copy()
T = 20.0
dt = 0.5
t_res = int(round(T / dt))
```

```python
modes = np.arange(1, 11)
a_initial_guess = np.array([
    0.69531099, 1.70119010, 3.70236071, 1.04948500, 0.45695289,
    -1.87686503, -1.91960996, -1.69153168, -0.42096132, 0.40649424
])

coefficients, H, f_final = optimizer(
    f_desired, a_initial_guess, modes, f0, x, v, T, t_res,
    tolerance=1e-3, learning_rate=0.15, max_iter=5000,
)
print("Final Coefficients:", coefficients)
```

```python
anim = plot_vp_solution(f0, x, v, T, t_res, H)
HTML(anim.to_jshtml())
```
